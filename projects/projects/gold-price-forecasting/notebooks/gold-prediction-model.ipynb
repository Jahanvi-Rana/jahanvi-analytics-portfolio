## 🚀 Core Analytical Workflows

### 1. Unified Data Engineering & Extraction
* **Multi-Sheet Conversion:** Automates the extraction of multi-sheet Excel data workbooks into structured, independent CSV training inputs.
* **Structural Auditing:** Conducts comprehensive data profile checks, dimension logging, and column vector type tracking.
* **Sanitization Pipeline:** Implements sequential data deduplication and handles time-series gaps via listwise index cleaning to ensure chronological continuity.

### 2. Multi-Target Advanced Feature Engineering
* **Localized Pricing Translation:** Programmatically transforms global commodity troy-ounce valuations into domestic Indian retail benchmarks ($10\text{g}$) using synchronized currency metrics.
* **Multi-Scale Momentum Mapping:** Computes Exponential Moving Averages ($\text{EMA}_3, \text{EMA}_6, \text{EMA}_{12}$) alongside active Rate of Change ($\text{ROC}$) and Momentum spreads to capture market velocities.
* **Volatility & Cyclical Profiles:** Traces market noise through rolling standard deviations and maps calendar seasonality using cyclical sine and cosine monthly coordinate transformations.

### 3. Progressive Machine Learning Ensembling
* **Anomalous Scaling:** Deploys robust feature scaling using `RobustScaler` to safely shield modeling datasets from global macroeconomic anomalies and extreme market spikes.
* **Chronological Validation:** Implements a strict time-series train/test split ($90\%$ training / $10\%$ testing) to eradicate data leakage during validation cycles.
* **Recency-Weighted Optimization:** Applies linear sample weight vectors over the training timeline to force the regressors to prioritize modern market regimes.
* **Blended Synthesis:** Fuses an optimized multi-model machine learning architecture ($40\%\ \text{XGBoost} + 30\%\ \text{LightGBM} + 30\%\ \text{CatBoost}$) to minimize individual variance.

### 4. Recursive Inference & Trend Correction
* **Dynamic Feedback Loops:** Executes an auto-regressive multi-step recursive forecasting sequence where predictions at step $T$ feed natively into the feature arrays for step $T+1$.
* **Variance Boundary Correction:** Integrates a calculated trend growth modifier based on recent data velocities to sustain forecasting accuracy over extended horizons.

# Step 4: US Price Prediction Pipeline (`V1_Core_Dataset`)

```python
# ==============================================================================
# PIPELINE INITIALIZATION: USD PRICE PREDICTION SYSTEM
# ==============================================================================

# [STEP 1] ENVIRONMENT PREPARATION: INSTALL REQUIRED EXTRA LIBRARIES
# Yandex's CatBoost framework requires explicit installation in typical notebook images.
!pip install catboost -q

# [STEP 2] CORE DEPENDENCIES & PREDICTIVE LIBRARIES IMPORT
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import time
import warnings

# Validation & Metric Tracking Utilities
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# Scaling: RobustScaler protects against massive market spikes/crashes in asset valuations
from sklearn.preprocessing import RobustScaler

# Ensemble Foundations: Gradient Boosting Tree Implementations
from xgboost import XGBRegressor      # Extreme Gradient Boosting Regressor
from lightgbm import LGBMRegressor    # Light Gradient Boosting Machine Regressor
from catboost import CatBoostRegressor  # CatBoost Regressor for robust categorical & structural modeling

# Suppress minor visual and optimization runtime warnings
warnings.filterwarnings('ignore')

# ==============================================================================
# VERIFY ENVIRONMENT & LOAD SOURCE DATASET
# ==============================================================================
print("="*70)
print("🚀 SYSTEM INITIALIZED: LOADING DATASET FOR USD PREDICTION")
print("="*70)

input_path = "/kaggle/working/V1_Cleaned_Gold_Dataset.csv"

if not os.path.exists(input_path):
    print(f"❌ Error: Dataset not found at {input_path}")
    print("Tip: Run your Excel-to-CSV conversion code first.")
else:
    df = pd.read_csv(input_path)

    # [STEP 3] CHRONOLOGICAL DATA ORDERING & BASE VARIABLE ASSIGNMENT
    # Ensure chronological validity to prevent analytical data leakage during modeling
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.sort_values('Date').reset_index(drop=True)
    
    # Defining our primary target benchmark variable in United States Dollars
    gold_col = 'Price_USD'

    # ==========================================================================
    # [STEP 4: FINANCIAL FEATURE ENGINEERING - USD MARKET INDICATORS]
    # ==========================================================================
    print("\n[STEP 1: FEATURE ENGINEERING]")
    print("-" * 30)
    print("Generating technical market indicators (EMA, Momentum, Volatility)...")
    
    # Trend Vectors: Exponential Moving Averages capturing short, mid, and long-term momentum shifts
    df['EMA_3'] = df[gold_col].ewm(span=3).mean()
    df['EMA_6'] = df[gold_col].ewm(span=6).mean()
    df['EMA_12'] = df[gold_col].ewm(span=12).mean()
    
    # Momentum & Velocity: Absolute and relative USD pricing variances over specific windows
    df['Momentum_1M'] = df[gold_col] - df['Gold_Lag_1M']
    df['Momentum_3M'] = df[gold_col] - df['Gold_Lag_3M']
    df['Volatility_3M'] = df[gold_col].rolling(3).std()
    df['Trend_Strength'] = df['EMA_3'] - df['EMA_12']
    df['ROC_1M'] = df[gold_col].pct_change(1)

    # Cyclical Calendar Configurations (Transforms calendar months into geometric sine/cosine coordinates)
    df['Month'] = df['Date'].dt.month
    df['Month_Sin'] = np.sin(2 * np.pi * df['Month'] / 12)
    df['Month_Cos'] = np.cos(2 * np.pi * df['Month'] / 12)

    # Sanitize edge null records resulting from calculated rolling steps
    df = df.dropna().reset_index(drop=True)

    # Separating Feature Matrix (X) from Target Labels (y)
    X = df.drop(columns=['Date', 'Next_Month_Gold_Price'])
    y = df['Next_Month_Gold_Price']

    # Normalize Feature Vectors using robust tracking mechanisms (Interquartile Outlier Management)
    scaler = RobustScaler()
    X_scaled = scaler.fit_transform(X)

    # Chronological Split (90% Train / 10% Test) to protect the sequencing of time elements
    split_index = int(len(df) * 0.90)
    X_train, X_test = X_scaled[:split_index], X_scaled[split_index:]
    y_train, y_test = y[:split_index], y[split_index:]

    print(f"✅ Data processed: {len(X_train)} training months, {len(X_test)} testing months.")

    # ==========================================================================
    # [STEP 5: PREDICTIVE ENSEMBLE TRAINING PHASE]
    # ==========================================================================
    print("\n[STEP 2: MODEL TRAINING PHASE]")
    print("-" * 30)
    print(">>> Training starts...")
    start_train = time.time()

    # Temporal Sequence Weighting: Linearly shifts importance up to 2x for modern/recent data trends
    sample_weights = np.linspace(1, 2, len(X_train))

    # Initialize the 3 Gradient Boosted Models with optimized parameter settings
    xgb_model = XGBRegressor(n_estimators=1000, learning_rate=0.01, max_depth=5, random_state=42)
    lgb_model = LGBMRegressor(n_estimators=800, learning_rate=0.01, max_depth=5, random_state=42, verbose=-1)
    cat_model = CatBoostRegressor(iterations=1000, learning_rate=0.01, depth=5, verbose=0, random_seed=42)

    # Fit the algorithms utilizing our localized custom timeline weights
    xgb_model.fit(X_train, y_train, sample_weight=sample_weights)
    lgb_model.fit(X_train, y_train, sample_weight=sample_weights)
    cat_model.fit(X_train, y_train, sample_weight=sample_weights)
    
    end_train = time.time()
    print(f">>> Training ends. (Time taken: {end_train - start_train:.2f} seconds)")

    
# ==========================================================================
    # [STEP 6: TESTING & OUT-OF-SAMPLE EVALUATION]
    # ==========================================================================
    print("\n[STEP 3: TESTING & PREDICTION PHASE]")
    print("-" * 30)
    print(">>> Testing starts...")
    start_test = time.time()  # Start testing timer
    
    # Generate standalone independent forecasts from each component model
    xgb_preds = xgb_model.predict(X_test)
    lgb_preds = lgb_model.predict(X_test)
    cat_preds = cat_model.predict(X_test)

    # Weighted Blend Calculation (Blended Ensemble Prediction)
    predictions = (0.40 * xgb_preds + 0.30 * lgb_preds + 0.30 * cat_preds)
    
    end_test = time.time()  # End testing timer
    print(f">>> Testing ends. (Time taken: {end_test - start_test:.4f} seconds)")

    # ==========================================================================
    # [STEP 7: PERFORMANCE METRIC COMPILATION - VALIDATION REPORT]
    # ==========================================================================
    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)
    mape = np.mean(np.abs((y_test - predictions) / y_test)) * 100
    accuracy = 100 - mape

    print("\n" + "="*70)
    print("USD MODEL PERFORMANCE REPORT")
    print("="*70)
    print(f"MAE (Mean Absolute Error)     : ${mae:.2f} USD")
    print(f"RMSE (Root Mean Squared Error): ${rmse:.2f} USD")
    print(f"R² Score (Goodness of Fit)    : {r2:.4f}")
    print(f"Prediction Accuracy Score     : {accuracy:.2f}%")
    print("="*70)
    

    # ==========================================================================
    # [STEP 8: AUTOREGRESSIVE FUTURE FORECASTING - NEXT 3 MONTHS OUTLOOK]
    # ==========================================================================
    print("\n[STEP 5: GENERATING 3-MONTH USD FORECAST]")
    print("-" * 30)
    
    future_predictions = []
    future_dates = []
    
    # Extract structural configuration from the terminal known index record
    latest_row = X.iloc[-1:].copy()
    current_date = df['Date'].max()

    # Stepwise dynamic looping calculation loop for multi-step predictions
    for i in range(3):
        latest_scaled = scaler.transform(latest_row)
        
        # Calculate localized prediction states across the model ensemble matrix
        p_xgb = xgb_model.predict(latest_scaled)[0]
        p_lgb = lgb_model.predict(latest_scaled)[0]
        p_cat = cat_model.predict(latest_scaled)[0]
        p_final = (0.4 * p_xgb + 0.3 * p_lgb + 0.3 * p_cat)
        
        # Trend adjustments: Embed a minute compounding growth constant factor (0.3% MoM)
        p_final = p_final * (1 + (0.003 * (i + 1)))
        
        future_predictions.append(p_final)
        current_date += pd.DateOffset(months=1)
        future_dates.append(current_date)
        
        # Auto-regressively shift calculated estimations forward as past features for the next iteration
        latest_row[gold_col] = p_final
        latest_row['Gold_Lag_1M'] = p_final

    # Build and print a clean tracking presentation table
    future_df = pd.DataFrame({'Future_Date': future_dates, 'Forecasted_Price_USD': future_predictions})
    print(future_df.to_string(index=False))

    # ==========================================================================
    # [STEP 9: MATPLOTLIB DATA VISUALIZATION PROJECTION]
    # ==========================================================================
    results_df = pd.DataFrame({
        'Date': df['Date'].iloc[split_index:].values, 
        'Actual': y_test.values, 
        'Predicted': predictions
    })

    plt.figure(figsize=(15, 7))
    
    # 1. Historical Data Layer Baseline
    plt.plot(df['Date'], df[gold_col], label='Historical Gold Price (USD)', color='gray', alpha=0.4)
    
    # 2. Out-of-sample Model Testing Evaluations Matrix
    plt.plot(results_df['Date'], results_df['Predicted'], label='Model Prediction (Test Set)', color='#1f77b4', linewidth=2)
    
    # 3. Future Projected Extension Window Horizon
    plt.plot(future_df['Future_Date'], future_df['Forecasted_Price_USD'], label='Future 3-Month Forecast', color='red', marker='o', linestyle='--', linewidth=2)
    
    # Presentation Configuration Tweaks
    plt.title("GOLD PRICE FORECASTING SYSTEM: HISTORICAL vs PREDICTED vs FUTURE (USD)", fontsize=16)
    plt.ylabel("USD Price ($)", fontsize=12)
    plt.xlabel("Timeline Year", fontsize=12)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("\n✅ USD FORECASTING PROJECT MODULE COMPLETED SUCCESSFULLY!")
```

```python
# ==============================================================================
# PIPELINE INITIALIZATION: DOMESTIC INDIAN GOLD PRICE PREDICTION SYSTEM
# ==============================================================================

# [STEP 1] ENVIRONMENT PREPARATION: DEPENDENCY CHECK & INSTALLATION
!pip install xgboost lightgbm catboost prophet -q

# [STEP 2] CORE DEPENDENCIES & COMPONENT PREDICTION LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
import os
import time

# Data Scaling Foundations for Multi-Model Matching
from sklearn.preprocessing import RobustScaler

# Model Validation Metrics
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Gradient Boosted Regression Core Architectures
from xgboost import XGBRegressor        
from lightgbm import LGBMRegressor      
from catboost import CatBoostRegressor  

# Suppress minor visual and optimization runtime warnings
warnings.filterwarnings('ignore')

# ==============================================================================
# VERIFY ENVIRONMENT & LOAD STANDARDIZED BASE DATASET
# ==============================================================================
print("="*80)
print("INDIAN GOLD PRICE FORECASTING SYSTEM")
print("="*80)

dataset_path = "/kaggle/working/V1_Cleaned_Gold_Dataset.csv"
df = pd.read_csv(dataset_path)

# [STEP 3] CHRONOLOGICAL TEMPORAL REALIGNMENT
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# ==============================================================================
# [STEP 4: CURRENCY CONVERSION & LOCAL MARKET FORMULATION]
# ==============================================================================
print("\nCreating Indian Gold Price...")

# Mathematically converting Global price (USD/Troy Ounce) to Domestic price (INR/10 Grams)
df['Gold_Price_INR_10g'] = (
    df['Price_USD'] *
    df['USDINR_Final'] *
    10
) / 31.1035

# ==============================================================================
# [STEP 5: SHIFT TARGET VALUE FOR AUTOREGRESSIVE SUPERVISED LEARNING]
# ==============================================================================
df['Next_Month_Gold_Price_INR'] = df['Gold_Price_INR_10g'].shift(-1)

# ==============================================================================
# [STEP 6: TECHNICAL FEATURE ENGINEERING - LOCAL INR VALUATION BASIS]
# ==============================================================================
print("\nGenerating Advanced Features...")
gold_col = 'Gold_Price_INR_10g'

# --- 6.1: EXPONENTIAL MOVING AVERAGES (EMA) ---
df['EMA_3'] = df[gold_col].ewm(span=3).mean()
df['EMA_6'] = df[gold_col].ewm(span=6).mean()
df['EMA_12'] = df[gold_col].ewm(span=12).mean()

# --- 6.2: ABSOLUTE MARKET MOMENTUM ---
df['Momentum_1M'] = df[gold_col] - df[gold_col].shift(1)
df['Momentum_3M'] = df[gold_col] - df[gold_col].shift(3)

# --- 6.3: HISTORICAL VOLATILITY TRACERS ---
df['Volatility_3M'] = df[gold_col].rolling(3).std()
df['Volatility_6M'] = df[gold_col].rolling(6).std()

# --- 6.4: PRICE RATE OF CHANGE (ROC) ---
df['ROC_1M'] = df[gold_col].pct_change(1)
df['ROC_3M'] = df[gold_col].pct_change(3)

# --- 6.5: TREND STRUCTURAL SIGNAL ---
df['Trend_Strength'] = df['EMA_3'] - df['EMA_12']

# --- 6.6: CALENDAR SEQUENCING FEATURES ---
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
df['Year'] = df['Date'].dt.year

# --- 6.7: CYCLICAL GEOMETRIC TIME ENCODING ---
df['Month_Sin'] = np.sin(2 * np.pi * df['Month'] / 12)
df['Month_Cos'] = np.cos(2 * np.pi * df['Month'] / 12)

# ==============================================================================
# [STEP 7: DATA CLEANING & RE-INDEXING VIA DROPPING BOUNDARY NULLS]
# ==============================================================================
df = df.dropna().reset_index(drop=True)
print(f"\nDataset Shape: {df.shape}")

# ==============================================================================
# [STEP 8: ISOLATE TRAINING MATRIX FEATURES FROM TARGET TARGET VALUES]
# ==============================================================================
target_column = 'Next_Month_Gold_Price_INR'
X = df.drop(columns=['Date', target_column])
y = df[target_column]

# ==============================================================================
# [STEP 9: ROBUST FEATURE NORMALIZATION]
# ==============================================================================
scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)

# ==============================================================================
# [STEP 10: CHRONOLOGICAL TIME SERIES TRAIN-TEST SPLIT]
# ==============================================================================
split_index = int(len(df) * 0.90)

X_train = X_scaled[:split_index]
X_test = X_scaled[split_index:]
y_train = y[:split_index]
y_test = y[split_index:]

# ==============================================================================
# [STEP 11: RECENT TREND TIME-WEIGHTING GENERATION]
# ==============================================================================
sample_weights = np.linspace(1, 2, len(X_train))

print(f"\nTraining Rows: {len(X_train)}")
print(f"Testing Rows: {len(X_test)}")

# ==============================================================================
# [STEP 12: HYPERPARAMETER ARCHITECTURE DEFINITIONS]
# ==============================================================================
print("\nTraining Models...")
start_train = time.time()

xgb_model = XGBRegressor(
    n_estimators=1200, learning_rate=0.01, max_depth=5,
    subsample=0.8, colsample_bytree=0.8, reg_alpha=0.5, reg_lambda=2,
    random_state=42
)

lgb_model = LGBMRegressor(
    n_estimators=1000, learning_rate=0.01, max_depth=5,
    num_leaves=31, min_child_samples=10, subsample=0.8, colsample_bytree=0.8,
    force_col_wise=True, random_state=42, verbose=-1
)

cat_model = CatBoostRegressor(
    iterations=1000, learning_rate=0.01, depth=5,
    loss_function='RMSE', verbose=0, random_seed=42
)

# ==============================================================================
# [STEP 13: PIPELINE EXECUTION: ENSEMBLE MODEL MODEL TRAINING]
# ==============================================================================
xgb_model.fit(X_train, y_train, sample_weight=sample_weights)
lgb_model.fit(X_train, y_train, sample_weight=sample_weights)
cat_model.fit(X_train, y_train, sample_weight=sample_weights)

end_train = time.time()
print(f"\nTraining Completed in {end_train - start_train:.2f} seconds")

# ==============================================================================
# [STEP 14: TESTING & INFERENCE EVALUATION METRIC SEGMENTATION]
# ==============================================================================
start_test = time.time()

xgb_preds = xgb_model.predict(X_test)
lgb_preds = lgb_model.predict(X_test)
cat_preds = cat_model.predict(X_test)

predictions = (0.40 * xgb_preds + 0.30 * lgb_preds + 0.30 * cat_preds)

# Post-Processing Short-Term Trend Drift Correction
recent_trend = df[gold_col].iloc[-1] / df[gold_col].iloc[-6]
predictions = predictions * (0.99 + (0.01 * recent_trend))

end_test = time.time()
print(f"Testing/Inference Completed in {end_test - start_test:.4f} seconds")

# ==============================================================================
# [STEP 15: PERFORMANCE METRICS COMPILATION & GENERATION]
# ==============================================================================
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)
mape = np.mean(np.abs((y_test - predictions) / y_test)) * 100
accuracy = 100 - mape

print("\n" + "="*80)
print("MODEL PERFORMANCE REPORT")
print("="*80)
print(f"\nMAE (Mean Absolute Error)     : ₹{mae:.2f}")
print(f"RMSE (Root Mean Squared Error): ₹{rmse:.2f}")
print(f"R² Score (Goodness of Fit)    : {r2:.4f}")
print(f"Prediction Accuracy Score     : {accuracy:.2f}%")
print("="*80)

# ==============================================================================
# [STEP 16: DATA FRAME MERGING & ERROR LOG COMPILATION]
# ==============================================================================
results_df = pd.DataFrame({
    'Date': df['Date'].iloc[split_index:].values,
    'Actual_Price_INR': y_test.values,
    'Predicted_Price_INR': predictions
})

results_df['Difference'] = results_df['Actual_Price_INR'] - results_df['Predicted_Price_INR']
results_df['Accuracy_%'] = 100 - (abs(results_df['Difference']) / results_df['Actual_Price_INR'] * 100)

print("\nACTUAL VS PREDICTED")
print(results_df.tail(15))

# ==============================================================================
# [STEP 17: AUTOREGRESSIVE RECURSIVE DEEP FORECASTING - HORIZON TIMELINES]
# ==============================================================================
print("\n" + "="*80)
print("INDIAN GOLD PRICE FORECAST : JUNE - AUGUST 2026")
print("="*80)

future_predictions = []
forecast_months = ['June 2026', 'July 2026', 'August 2026']
future_dates = pd.to_datetime(["2026-06-01", "2026-07-01", "2026-08-01"])

latest_row = X.iloc[-1:].copy()

for i in range(3):
    latest_scaled = scaler.transform(latest_row)

    pred_xgb = xgb_model.predict(latest_scaled)[0]
    pred_lgb = lgb_model.predict(latest_scaled)[0]
    pred_cat = cat_model.predict(latest_scaled)[0]

    final_prediction = (0.40 * pred_xgb + 0.30 * pred_lgb + 0.30 * pred_cat)

    growth_factor = (1 + (0.007 * (i + 1)))
    final_prediction = final_prediction * growth_factor

    future_predictions.append(final_prediction)

    # Feedback Loop Logic
    previous_price = latest_row[gold_col].values[0]
    latest_row[gold_col] = final_prediction

    latest_row['EMA_3'] = 0.7 * final_prediction + 0.3 * previous_price
    latest_row['EMA_6'] = 0.5 * final_prediction + 0.5 * previous_price
    latest_row['EMA_12'] = 0.3 * final_prediction + 0.7 * previous_price

    latest_row['Momentum_1M'] = final_prediction - previous_price
    latest_row['Momentum_3M'] = final_prediction - previous_price

    latest_row['Volatility_3M'] = abs(final_prediction - previous_price)
    latest_row['Volatility_6M'] = abs(final_prediction - previous_price)

    latest_row['ROC_1M'] = (final_prediction - previous_price) / previous_price
    latest_row['ROC_3M'] = (final_prediction - previous_price) / previous_price

# ==============================================================================
# [STEP 18: PRESENTATION REPORT FORMATTING]
# ==============================================================================
future_df = pd.DataFrame({
    'Forecast_Month': forecast_months,
    'Forecast_Date': future_dates,
    'Predicted_Indian_Gold_Price_INR_10g': np.round(future_predictions, 2)
})

print("\n")
print("="*80)
print("FUTURE FORECAST RESULTS")
print("="*80)
print(future_df.to_string(index=False))

# ==============================================================================
# [STEP 19: MATPLOTLIB ADVANCED PRESENTATION VISUALIZATION]
# ==============================================================================
plt.figure(figsize=(24, 10))

plt.plot(df['Date'], df[gold_col], label='Historical Indian Gold Price', linewidth=3, color='black')
plt.plot(results_df['Date'], results_df['Predicted_Price_INR'], label='Model Predictions', linestyle='--', linewidth=3, color='blue')
plt.plot(future_df['Forecast_Date'], future_df['Predicted_Indian_Gold_Price_INR_10g'], marker='o', markersize=12, linewidth=4, linestyle='--', color='red', label='Forecast : June-August 2026')

for x, y in zip(future_df['Forecast_Date'], future_df['Predicted_Indian_Gold_Price_INR_10g']):
    plt.text(x, y + 300, f'₹{y:,.0f}', fontsize=11, fontweight='bold')

plt.axvspan(future_df['Forecast_Date'].iloc[0], future_df['Forecast_Date'].iloc[-1], alpha=0.1, color='red')
plt.axvline(x=results_df['Date'].iloc[0], linestyle='--', color='green', linewidth=2)

plt.title('INDIAN GOLD PRICE FORECAST : JUNE-AUGUST 2026', fontsize=24, fontweight='bold')
plt.xlabel('Date', fontsize=16)
plt.ylabel('Indian Gold Price (INR per 10g)', fontsize=16)
plt.legend(fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ==============================================================================
# [STEP 20: SYSTEM COMPLETION SIGNOFF CONSOLE TRACES]
# ==============================================================================
print("\n" + "="*80)
print("FORECASTING COMPLETED SUCCESSFULLY")
print("="*80)

print("\nForecast Months Generated:")
for month, value in zip(forecast_months, future_predictions):
    print(f"{month} : ₹{value:,.2f}")
```
